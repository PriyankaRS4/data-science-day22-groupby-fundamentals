# Task 22 — GroupBy Fundamentals
Titanic dataset GroupBy practice.

```python
import pandas as pd
df=pd.read_csv('titanic.csv')
for c in ['pclass','survived','age','fare']: df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
# 1. Sum
df.groupby('pclass')['fare'].sum()
```

```python
# 2. Mean
df.groupby('pclass')['fare'].mean()
```

```python
# 3. Count
df.groupby('sex')['survived'].count()
```

```python
# 4. Minimum and maximum
df.groupby('embarked')['fare'].agg(['min','max'])
```

```python
# 5. Multiple aggregations
df.groupby('pclass').agg(passengers=('pclass','count'),total_fare=('fare','sum'),average_fare=('fare','mean'),min_fare=('fare','min'),max_fare=('fare','max'))
```

```python
# 6. Multiple-column grouping
df.groupby(['sex','pclass']).agg(passengers=('survived','count'),survivors=('survived','sum'),average_fare=('fare','mean')).reset_index()
```

## Short Analysis
GroupBy allows numerical measures to be summarized across categories such as passenger class, sex and embarkation port. Multiple columns can also be grouped together for more detailed analysis.
